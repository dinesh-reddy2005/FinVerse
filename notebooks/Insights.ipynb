```python
import mysql.connector
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from IPython.display import display

# Display settings
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)

# Plot style
sns.set_theme(style="whitegrid")

print("Libraries imported successfully.")
```

```python
customers = pd.read_sql(
    "SELECT * FROM customers",
    conn
)

print("Customers:", customers.shape)
display(customers.head())
```

```python
transactions = pd.read_sql(
    "SELECT * FROM transactions",
    conn
)

print("Transactions:", transactions.shape)
display(transactions.head())
```

```python
merchants = pd.read_sql(
    "SELECT * FROM merchants",
    conn
)

print("Merchants:", merchants.shape)
display(merchants.head())
```

```python
branches = pd.read_sql(
    "SELECT * FROM branches",
    conn
)

print("Branches:", branches.shape)
display(branches.head())
```

```python
transactions['transaction_date'] = pd.to_datetime(
    transactions['transaction_date'],
    errors='coerce'
)

print(transactions['transaction_date'].min())
print(transactions['transaction_date'].max())
```

```python
transactions['date'] = transactions['transaction_date'].dt.date
transactions['year'] = transactions['transaction_date'].dt.year
transactions['month'] = transactions['transaction_date'].dt.month
transactions['month_name'] = transactions['transaction_date'].dt.month_name()
transactions['day_name'] = transactions['transaction_date'].dt.day_name()
```

```python
total_customers = customers['customer_id'].nunique()

total_transactions = transactions['transaction_id'].nunique()

total_transaction_value = transactions['amount'].sum()

average_transaction = transactions['amount'].mean()

fraud_transactions = (
    transactions['is_fraud'] == 1
).sum()

fraud_amount = transactions.loc[
    transactions['is_fraud'] == 1,
    'amount'
].sum()

fraud_rate = (
    fraud_transactions / total_transactions
) * 100

total_merchants = merchants['merchant_id'].nunique()

total_branches = branches['branch_id'].nunique()

print("========== FINTECH KPIs ==========")

print(f"Total Customers       : {total_customers:,}")
print(f"Total Transactions    : {total_transactions:,}")
print(f"Total Transaction ₹   : {total_transaction_value:,.2f}")
print(f"Average Transaction ₹ : {average_transaction:,.2f}")
print(f"Fraud Transactions    : {fraud_transactions:,}")
print(f"Fraud Amount ₹        : {fraud_amount:,.2f}")
print(f"Fraud Rate            : {fraud_rate:.2f}%")
print(f"Total Merchants       : {total_merchants:,}")
print(f"Total Branches        : {total_branches:,}")
```

```python
## Top customers by transaction value
```

```python
customer_insights = (
    transactions
    .groupby('customer_id')
    .agg(
        transaction_count=('transaction_id', 'count'),
        total_transaction_value=('amount', 'sum'),
        average_transaction_value=('amount', 'mean'),
        fraud_transactions=('is_fraud', 'sum')
    )
    .reset_index()
)

customer_insights = customer_insights.merge(
    customers[['customer_id', 'customer_name']],
    on='customer_id',
    how='left'
)

customer_insights = customer_insights.sort_values(
    'total_transaction_value',
    ascending=False
)

display(customer_insights.head(10))
```

```python
## Fraud by transaction type
```

```python
fraud_by_type = (
    transactions
    .groupby('transaction_type')
    .agg(
        total_transactions=('transaction_id', 'count'),
        fraud_transactions=('is_fraud', 'sum')
    )
    .reset_index()
)

fraud_by_type['fraud_rate'] = (
    fraud_by_type['fraud_transactions']
    / fraud_by_type['total_transactions']
    * 100
)

display(
    fraud_by_type.sort_values(
        'fraud_rate',
        ascending=False
    )
)
```

```python
plt.figure(figsize=(10, 5))

sns.barplot(
    data=fraud_by_type,
    x='transaction_type',
    y='fraud_rate'
)

plt.title("Fraud Rate by Transaction Type")
plt.xlabel("Transaction Type")
plt.ylabel("Fraud Rate (%)")

plt.xticks(rotation=45)

plt.show()
```

```python
## Fraud by state
```

```python
transaction_customer = transactions.merge(
    customers[['customer_id', 'customer_name', 'state', 'city', 'branch_id']],
    on='customer_id',
    how='left'
)
```

```python
fraud_by_state = (
    transaction_customer
    .groupby('state')
    .agg(
        total_transactions=('transaction_id', 'count'),
        fraud_transactions=('is_fraud', 'sum'),
        total_amount=('amount', 'sum'),
        fraud_amount=('amount',
                         lambda x: x[
                             transaction_customer.loc[x.index, 'is_fraud'] == 1
                         ].sum())
    )
    .reset_index()
)

fraud_by_state['fraud_rate'] = (
    fraud_by_state['fraud_transactions']
    / fraud_by_state['total_transactions']
    * 100
)

display(
    fraud_by_state.sort_values(
        'fraud_rate',
        ascending=False
    )
)
```

```python
top_states = fraud_by_state.sort_values(
    'fraud_rate',
    ascending=False
).head(10)

plt.figure(figsize=(12, 6))

sns.barplot(
    data=top_states,
    x='fraud_rate',
    y='state'
)

plt.title("Top States by Fraud Rate")
plt.xlabel("Fraud Rate (%)")
plt.ylabel("State")

plt.show()
```

```python
## Branch analysis
```

```python
transaction_branch = (
    transactions
    .merge(
        customers[['customer_id', 'branch_id']],
        on='customer_id',
        how='left'
    )
    .merge(
        branches,
        on='branch_id',
        how='left'
    )
)

branch_analysis = (
    transaction_branch
    .groupby(['branch_id', 'bank_branch', 'state'])
    .agg(
        transaction_count=('transaction_id', 'count'),
        total_amount=('amount', 'sum'),
        average_amount=('amount', 'mean'),
        fraud_transactions=('is_fraud', 'sum')
    )
    .reset_index()
)

branch_analysis['fraud_rate'] = (
    branch_analysis['fraud_transactions']
    / branch_analysis['transaction_count']
    * 100
)

display(
    branch_analysis.sort_values(
        'total_amount',
        ascending=False
    ).head(20)
)
```

```python
## Customer Risk Score
```

```python
risk_data = (
    transactions
    .groupby('customer_id')
    .agg(
        total_transactions=('transaction_id', 'count'),
        fraud_transactions=('is_fraud', 'sum'),
        average_transaction=('amount', 'mean'),
        total_amount=('amount', 'sum')
    )
    .reset_index()
)

risk_data = risk_data.merge(
    customers[['customer_id', 'customer_name']],
    on='customer_id',
    how='left'
)
```

```python
risk_data['risk_score'] = 0

# Fraud history
risk_data.loc[
    risk_data['fraud_transactions'] >= 3,
    'risk_score'
] += 40

risk_data.loc[
    (risk_data['fraud_transactions'] >= 1) &
    (risk_data['fraud_transactions'] < 3),
    'risk_score'
] += 25

# High transaction amount
risk_data.loc[
    risk_data['average_transaction'] > 50000,
    'risk_score'
] += 30

risk_data.loc[
    (risk_data['average_transaction'] > 25000) &
    (risk_data['average_transaction'] <= 50000),
    'risk_score'
] += 20

risk_data.loc[
    (risk_data['average_transaction'] > 10000) &
    (risk_data['average_transaction'] <= 25000),
    'risk_score'
] += 10

# Transaction frequency
risk_data.loc[
    risk_data['total_transactions'] > 100,
    'risk_score'
] += 20

risk_data.loc[
    (risk_data['total_transactions'] > 50) &
    (risk_data['total_transactions'] <= 100),
    'risk_score'
] += 10
```

```python
insights = []

# Fraud
if fraud_rate > 2:
    insights.append(
        f"Fraud rate is {fraud_rate:.2f}%, which requires attention."
    )
else:
    insights.append(
        f"Fraud rate is {fraud_rate:.2f}%."
    )

# Transaction value
insights.append(
    f"Total transaction value is ₹{total_transaction_value:,.2f}."
)

# Average transaction
insights.append(
    f"Average transaction value is ₹{average_transaction:,.2f}."
)

# Highest fraud state
if not fraud_by_state.empty:
    highest_fraud_state = fraud_by_state.iloc[
        fraud_by_state['fraud_rate'].argmax()
    ]

    insights.append(
        f"{highest_fraud_state['state']} has the highest observed "
        f"state-level fraud rate of "
        f"{highest_fraud_state['fraud_rate']:.2f}%."
    )

# Highest fraud merchant category
if not merchant_category_analysis.empty:
    highest_fraud_category = merchant_category_analysis.iloc[
        merchant_category_analysis['fraud_rate'].argmax()
    ]

    insights.append(
        f"{highest_fraud_category['merchant_category']} has the highest "
        f"merchant-category fraud rate of "
        f"{highest_fraud_category['fraud_rate']:.2f}%."
    )


print("========== GENERATED INSIGHTS ==========\n")

for i, insight in enumerate(insights, 1):
    print(f"{i}. {insight}")
```
