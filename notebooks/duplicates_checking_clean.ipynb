```python
import pandas as pd

df = pd.read_csv("customers.csv")

print("Original rows:", len(df))

# Remove completely empty rows
df = df.dropna(how="all")

# Remove duplicate customer IDs
df = df.drop_duplicates(subset=["customer_id"], keep="first")

# Remove records where customer_id is missing
df = df.dropna(subset=["customer_id"])

print("Clean rows:", len(df))

print("Unique customer IDs:", df["customer_id"].nunique())

df.to_csv("customers_cleaned.csv", index=False)

print("Cleaned file created successfully.")
```

```python
import pandas as pd
import re

# Load file
df = pd.read_csv("merchants.csv")

print("========== BASIC INFORMATION ==========")

print("Total rows:", len(df))
print("Total columns:", len(df.columns))

print("\nColumns:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)


# ==========================================
# 1. MISSING VALUES
# ==========================================

print("\n========== MISSING VALUES ==========")

print(df.isnull().sum())


# ==========================================
# 2. DUPLICATE ROWS
# ==========================================

print("\n========== DUPLICATES ==========")

print("Duplicate complete rows:",
      df.duplicated().sum())


# ==========================================
# 3. DUPLICATE MERCHANT IDs
# ==========================================

print("Duplicate merchant IDs:",
      df["merchant_id"].duplicated().sum())


# Show duplicate merchant IDs if any
duplicates = df[
    df["merchant_id"].duplicated(keep=False)
]

if len(duplicates) > 0:
    print("\nDuplicate merchant records:")
    print(duplicates)


# ==========================================
# 4. BLANK VALUES
# ==========================================

print("\n========== BLANK VALUES ==========")

for column in df.columns:
    blank_count = (
        df[column]
        .astype(str)
        .str.strip()
        .eq("")
        .sum()
    )

    print(column, ":", blank_count)


# ==========================================
# 5. LEADING / TRAILING SPACES
# ==========================================

print("\n========== WHITESPACE CHECK ==========")

for column in df.columns:

    whitespace_count = (
        df[column].astype(str)
        != df[column].astype(str).str.strip()
    ).sum()

    print(column, ":", whitespace_count)


# ==========================================
# 6. MERCHANT ID UUID VALIDATION
# ==========================================

print("\n========== UUID VALIDATION ==========")

uuid_pattern = (
    r"^[0-9a-fA-F]{8}-"
    r"[0-9a-fA-F]{4}-"
    r"[1-5][0-9a-fA-F]{3}-"
    r"[89abAB][0-9a-fA-F]{3}-"
    r"[0-9a-fA-F]{12}$"
)

invalid_uuid = ~df["merchant_id"].astype(str).str.match(
    uuid_pattern
)

print("Invalid merchant IDs:", invalid_uuid.sum())

if invalid_uuid.sum() > 0:
    print(df.loc[invalid_uuid, "merchant_id"])


# ==========================================
# 7. MERCHANT CATEGORIES
# ==========================================

print("\n========== MERCHANT CATEGORIES ==========")

print("Number of categories:",
      df["merchant_category"].nunique())

print("\nCategory counts:")
print(df["merchant_category"].value_counts())


# ==========================================
# 8. FINAL SUMMARY
# ==========================================

print("\n========== FINAL SUMMARY ==========")

print("Total rows:", len(df))
print("Missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())
print("Duplicate merchant IDs:",
      df["merchant_id"].duplicated().sum())
print("Invalid UUIDs:", invalid_uuid.sum())

print("\nData quality check completed.")
```
