```python
import pandas as pd

file_path = "customers_uncleaned.csv"

df = pd.read_csv(file_path, dtype=str, keep_default_na=False)

print("=" * 60)
print("CUSTOMER DATA QUALITY REPORT")
print("=" * 60)

print("\n1. Total rows:")
print(len(df))

print("\n2. Columns:")
print(df.columns.tolist())

# Duplicate complete rows
print("\n3. Duplicate complete rows:")
print(df.duplicated().sum())

# Duplicate customer IDs
print("\n4. Duplicate customer IDs:")
print(df["customer_id"].duplicated().sum())

# Unique customer IDs
print("\n5. Unique customer IDs:")
print(df["customer_id"].nunique())

# Missing values
print("\n6. Missing / blank values:")
for col in df.columns:
    count = (df[col].str.strip() == "").sum()
    print(f"{col}: {count}")

# Account types
print("\n7. Account types:")
print(df["account_type"].value_counts(dropna=False))

# Gender
print("\n8. Gender values:")
print(df["gender"].value_counts(dropna=False))

# States
print("\n9. Suspicious state values:")
print(df[df["state"].str.strip() == "."])

# Age
age = pd.to_numeric(df["age"], errors="coerce")

print("\n10. Age information:")
print("Non-numeric ages:", age.isna().sum())
print("Minimum age:", age.min())
print("Maximum age:", age.max())
print("Age below 18:", (age < 18).sum())

# Branch IDs
branch = pd.to_numeric(df["branch_id"], errors="coerce")

print("\n11. Branch ID information:")
print("Non-numeric branch IDs:", branch.isna().sum())
print("Minimum branch ID:", branch.min())
print("Maximum branch ID:", branch.max())
print("Invalid branch IDs:", ((branch < 1) | (branch > 149)).sum())

# Whitespace
print("\n12. Leading/trailing spaces:")
for col in df.columns:
    count = (df[col] != df[col].str.strip()).sum()
    if count > 0:
        print(f"{col}: {count}")
```

```python
noise_mask = (
    df["customer_id"].str.match(r"^999881974\d+$", na=False)
    |
    df["customer_name"].str.match(r"^Noise Customer \d+$", na=False)
)

noise_rows = df[noise_mask]

print("Noise rows:", len(noise_rows))

print("\nFirst 10 noise rows:")
print(noise_rows.head(10))
```

```python
age = pd.to_numeric(df["age"], errors="coerce")

print("Invalid age:", age.isna().sum())
print("Age < 18:", (age < 18).sum())
print("Age > 100:", (age > 100).sum())
```

```python
import pandas as pd

df = pd.read_csv(
    "branches_raw_uncleaned.csv",
    dtype=str,
    keep_default_na=False
)

print("=" * 50)
print("DATA QUALITY CHECK")
print("=" * 50)

# Total rows
print("\nTotal rows:", len(df))

# Duplicate complete rows
print(
    "Duplicate complete rows:",
    df.duplicated().sum()
)

# Duplicate branch IDs
print(
    "Duplicate branch IDs:",
    df["branch_id"].duplicated().sum()
)

# Display duplicate branch IDs
print("\nDuplicate branch records:")
print(
    df[df["branch_id"].duplicated(keep=False)]
)

# Missing/blank values
print("\nMissing / blank values:")

for column in df.columns:
    count = (
        df[column]
        .astype(str)
        .str.strip()
        .eq("")
        .sum()
    )

    print(f"{column}: {count}")
```

```python
import pandas as pd

df = pd.read_csv(
    "merchants_raw_uncleaned.csv",
    dtype=str,
    keep_default_na=False
)

print("=" * 60)
print("MERCHANT DATA QUALITY CHECK")
print("=" * 60)

# Total rows
print("\nTotal rows:")
print(len(df))

# Columns
print("\nColumns:")
print(df.columns.tolist())

# Duplicate complete rows
print("\nDuplicate complete rows:")
print(df.duplicated().sum())

# Duplicate merchant IDs
print("\nDuplicate merchant IDs:")
print(df["merchant_id"].duplicated().sum())

# Unique merchant IDs
print("\nUnique merchant IDs:")
print(df["merchant_id"].nunique())

# Missing / blank values
print("\nBlank values:")

for column in df.columns:
    count = (
        df[column]
        .astype(str)
        .str.strip()
        .eq("")
        .sum()
    )

    print(f"{column}: {count}")
```

```python
duplicates = df[
    df["merchant_id"].duplicated(keep=False)
]

print(duplicates)
```

```python
duplicate_counts = (
    df["merchant_id"]
    .value_counts()
    .loc[lambda x: x > 1]
)

print(duplicate_counts)
```
